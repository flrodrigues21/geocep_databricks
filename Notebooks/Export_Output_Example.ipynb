```python
# Origem no Volume
origem = "/Volumes/geocep/default/geocep/output/historico/ano=2026/mes=11"

# Destino no repositório Git
destino = "/Workspace/Users/flrodrigues21@gmail.com/geocep_databricks/output/ano=2026/mes=11"

# Cria a pasta de destino
import os
os.makedirs(destino, exist_ok=True)

# Encontra um arquivo Parquet
arquivos = [
    f.path for f in dbutils.fs.ls(origem)
    if f.name.endswith(".parquet")
]

# Copia um arquivo
if arquivos:
    dbutils.fs.cp(
        arquivos[0],
        "file:" + destino + "/" + arquivos[0].split("/")[-1]
    )
    print("Parquet copiado para o Git!")
else:
    print("Nenhum Parquet encontrado.")
```
