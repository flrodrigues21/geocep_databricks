```python
import os
import shutil
from pathlib import Path

# Histórico já gerado no Databricks
origem = Path(
    "/Volumes/geocep/default/geocep/output/historico"
)

# Localiza a raiz do repositório a partir do notebook
repo = Path(os.getcwd()).parent

print("Repositório identificado:", repo)

if repo.name != "geocep_databricks":
    raise ValueError("Confirme o caminho da Git folder antes de continuar.")

# Localiza um arquivo Parquet do histórico
arquivos = sorted(origem.glob("ano=*/mes=*/*.parquet"))

if not arquivos:
    raise FileNotFoundError("Nenhum Parquet encontrado no histórico.")

arquivo = arquivos[-1]

# Preserva a estrutura de particionamento
destino = repo / "output" / arquivo.relative_to(origem)
destino.parent.mkdir(parents=True, exist_ok=True)

# Copia o arquivo, sem alterar o histórico original
shutil.copy2(arquivo, destino)

print("Amostra copiada para:", destino)
```
